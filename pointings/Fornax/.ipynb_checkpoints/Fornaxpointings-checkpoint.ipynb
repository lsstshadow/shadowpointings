```python
#this merges the galaxy and fornax pointings catalogs into a file that can be displayed by the legacy viewer:
#https://www.legacysurvey.org/viewer

df1 = pd.read_csv('output.csv')
df2 = pd.read_csv('yo.csv',header=None)
df2.columns = df1.columns
target_column = 'name'
specific_value = 'gal'
new_value = ''  # Empty string

df2.loc[df2[target_column] == specific_value, target_column] = new_value


concat_df_rows = pd.concat([df1,df2],ignore_index=True)

concat_df_rows.to_csv('fornax_output.csv',index=False)
```
